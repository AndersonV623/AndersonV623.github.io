```python
import json
import pandas as pd
from pandas import json_normalize
URL = "C:\\Users\\jawar\\Downloads\\SIMCHS_2026_03 (4).json"
```

```python
# Cargar el archivo JSON
with open(URL, "r", encoding="utf-8") as f:
    data = json.load(f)

# Normalizar: expandir la lista de 'respuestas' y traer los metadatos
df = pd.json_normalize(
    data,
    record_path=["respuestas"],   # lista que queremos expandir
    meta=[
        "id",
        "estado",
        ["tiempo", "inicio"],
        ["tiempo", "fin"],
        ["tiempo", "duracionSegundos"],
        ["progreso", "respondidas"],
        ["progreso", "total"],
        ["progreso", "porcentaje"],
        ["datosGenerales", "rangoEdad"],
        ["datosGenerales", "genero"],
        ["datosGenerales", "nivelEducativo"],
        ["datosGenerales", "tipoSalud"],
        ["datosGenerales", "municipio"],
        ["datosGenerales", "musica"]
    ]
)
```

```python
for index, row in df[["item", "valor"]].iterrows():
    print(f"{row['valor']} -resp-:-item- {row['item']}")
```
