# **Resumen**
<p style="text-align:justify">El proyecto SIMCH-S (Sistema Integral Multidimensional de Conciencia Humana en la Sociedad) propone el diseño, validación e implementación de un modelo científico para la medición estructural de la conciencia humana desde una perspectiva tridimensional. El modelo evalúa quince dimensiones fundamentales —entre ellas cognición, emoción, autoconciencia, regulación emocional, propósito y coherencia interna— las cuales se integran en tres macrofactores estructurales: Cognitivo (N1), Emocional (N2) y Conductual (N3).

<p style="text-align:justify">El desarrollo del proyecto se organiza en cuatro fases: (1) formalización académica y fundamentación teórica, (2) construcción del modelo científico mediante operacionalización de variables y definición de constructos, (3) diseño y validación psicométrica del instrumento de medición, y (4) implementación de un sistema digital capaz de procesar datos, generar perfiles tridimensionales (R3) y ofrecer retroalimentación estructural individual y social.

<p style="text-align:justify">El instrumento, compuesto por 150 ítems distribuidos en 15 dimensiones, permite no solo evaluar el estado estructural de conciencia en una persona, sino también realizar seguimiento longitudinal a través de una plataforma web con control de usuarios. Esto posibilita visualizar la evolución individual y analizar tendencias colectivas de manera agregada y anonimizada.

<p style="text-align:justify">SIMCH-S busca aportar una arquitectura científica que integre medición psicológica, representación tridimensional y procesamiento digital, orientada al desarrollo personal, la investigación académica y el análisis estructural del comportamiento humano en contextos educativos y sociales.

## Marco teórico

<p style="text-align:justify">El marco teórico del sistema se fundamenta en un enfoque interdisciplinar que integra múltiples corrientes científicas y humanistas, permitiendo una comprensión integral del ser humano como sistema complejo, consciente y social.

### 1. Psicología humanista

<p style="text-align:justify">Basada en los aportados de Abraham Maslow y Carl Rogers, la psicología humanista concibe al ser humano como un ser orientado al crecimiento, la autorrealización y el desarrollo del potencial interno.

- Este proyecto adopta la visión del ser humano como un agente activo de su propio desarrollo.

### 2. Psicología positiva

<p style="text-align:justify">Desde los trabajos de Martin Seligman, se integra el enfoque en fortalezas humanas, bienestar, propósito, resiliencia y sentido de vida, entendiendo el desarrollo humano como expansión de capacidades internas y no solo como corrección de déficits.

### 3. Psicología social

<p style="text-align:justify">Se fundamenta en la comprensión del individuo como ser relacional, influido por normas sociales, cultura, contexto y dinámicas grupales. La identidad humana se construye socialmente.

### 4. Psicología cognitiva

<p style="text-align:justify">Aporta el estudio de los procesos mentales: percepción, atención, memoria, pensamiento, toma de decisiones y construcción de significado, fundamentales para comprender la conciencia y la autoconciencia.

### 5. Neurociencia

<p style="text-align:justify">Contribuye a la comprensión biológica de la emoción, la conciencia, la regulación emocional y los procesos cognitivos, permitiendo un enfoque integrador mente–cerebro–conducta.

### 6. Psicología del desarrollo

<p style="text-align:justify">Desde Erik Erikson y Jean Piaget, se integra la visión del desarrollo humano como proceso evolutivo continuo a lo largo del ciclo vital.

### 7. Psicología existencial

<p style="text-align:justify">Aporta los conceptos de sentido, propósito, responsabilidad personal, libertad y construcción de identidad, fundamentales para las dimensiones de propósito y trascendencia.

### 8. Teoría de sistemas

<p style="text-align: justify">Desde el enfoque sistémico, el ser humano se comprende como un sistema abierto, dinámico, interdependiente y adaptativo, en constante interacción con otros sistemas (sociales, culturales, simbólicos).

### 9. Constructivismo

<p style="text-align: justify">La realidad es entendida como una construcción subjetiva mediada por la experiencia, la cultura y la cognición. El sistema no mide realidades absolutas, sino realidades interpretadas.

### 10. Enfoque de sistemas complejos

<p style="text-align: justify">El ser humano y la sociedad son considerados sistemas complejos no lineales, donde emergen patrones, propiedades colectivas y dinámicas no predecibles.

### 11. Teoría de la conciencia

<p style="text-align: justify">Se integra la conciencia como un proceso funcional de integración cognitiva, emocional y social, no como un fenómeno metafísico.

### 12. Ética del desarrollo humano

<p style="text-align: justify">El proyecto se fundamenta en principios de dignidad humana, autonomía, respeto, libertad individual, no manipulación y responsabilidad social.

<p style="text-align: justify">Este marco teórico permite sustentar científicamente el sistema como un modelo legítimo de investigación humana aplicada, integrando ciencia, ética y desarrollo social.

## Planteamiento del problema

<p style="text-align: justify"> En la actualidad, los sistemas de medición del ser humano se encuentran fragmentados en dimensiones aisladas: cognición, emoción, personalidad, bienestar, conducta social, desempeño, entre otras. La mayoría de los instrumentos existentes evalúan aspectos parciales del individuo, sin una integración estructural de la percepción, la autoconciencia y la conciencia social como un sistema unificado.

<p style="text-align: justify">Esta fragmentación genera una comprensión incompleta del desarrollo humano, limitando la posibilidad de construir modelos integrales que expliquen cómo los estados internos del individuo influyen en su comportamiento social, en su relación con los demás y en la calidad del tejido social.

<p style="text-align: justify">Adicionalmente, los sistemas actuales de evaluación humana suelen estar orientados a fines clínicos, diagnósticos, laborales o de selección de personal, dejando un vacío en modelos de medición éticos, no clínicos y no instrumentales, centrados en el desarrollo consciente del ser humano como sujeto social.

<p style="text-align: justify">No existe actualmente un sistema integrado que permita:

- Medir percepción del mundo
- Medir autoconciencia
- Medir conciencia social
- Analizar patrones humanos
- Generar retroalimentación reflexiva
- Promover desarrollo humano

<p style="text-align: justify">Desde una perspectiva social, esta carencia limita la posibilidad de diseñar intervenciones formativas, educativas y culturales basadas en datos reales sobre estados internos humanos, percepción social y conciencia colectiva.

<p style="text-align: justify">Por tanto, se identifica la necesidad de diseñar un sistema integral, ético y científico que permita medir, comprender y acompañar el desarrollo humano en su dimensión individual y social.

## 2. Definición de Constructos

<p style="text-align: justify">La definición de constructos establece el marco conceptual científico del sistema SIMCH-S, delimitando semántica, teórica y metodológicamente cada conciencia como una entidad psicológica medible. Cada constructo se define como una variable latente que se manifiesta a través de indicadores observables.

#### Constructo 1: Autoconciencia

<p style="text-align: justify">Capacidad del individuo para percibirse, reconocerse y comprenderse a sí mismo como entidad diferenciada, integrando procesos emocionales, cognitivos e identitarios en un sistema coherente de autoobservación consciente.

#### Constructo 2: Conciencia emocional

<p style="text-align: justify">Capacidad de reconocer, comprender, regular y expresar los estados emocionales propios de forma adaptativa y funcional dentro del contexto social.

#### Constructo 3: Conciencia social

<p style="text-align: justify">Capacidad de percibir, interpretar y responder de manera consciente al entorno social, integrando normas, dinámicas interpersonales y responsabilidad colectiva.

#### Constructo 4: Conciencia empática

<p style="text-align: justify">Capacidad de comprender y resonar con los estados emocionales, cognitivos y experienciales de otras personas, manteniendo diferenciación del yo.

#### Constructo 5: Conciencia ética

<p style="text-align: justify">Sistema interno de valores, principios y normas morales que guían la conducta, la toma de decisiones y la autorregulación comportamental.

#### Constructo 6: Conciencia crítica

<p style="text-align: justify">Capacidad de análisis reflexivo, pensamiento autónomo, evaluación racional de la información y metacognición aplicada a la realidad personal y social.

#### Constructo 7: Conciencia existencial

<p style="text-align: justify">Capacidad de otorgar significado a la propia existencia, construir sentido vital y elaborar una narrativa coherente de propósito y trascendencia personal.

#### Constructo 8: Conciencia corporal

<p style="text-align: justify">Capacidad de integración entre procesos mentales, emocionales y fisiológicos mediante la percepción consciente del cuerpo y sus señales.

#### Constructo 9: Conciencia comunicativa

<p style="text-align: justify">Capacidad de expresar, escuchar y comprender información emocional, cognitiva y social de manera consciente, empática y estructurada.

#### Constructo 10: Conciencia relacional

<p style="text-align: justify">Capacidad de construir, mantener y regular vínculos humanos estables, seguros y funcionales en contextos interpersonales.

#### Constructo 11: Conciencia adaptativa

<p style="text-align: justify">Capacidad de ajuste psicológico, emocional y conductual frente a cambios, crisis y contextos complejos.

#### Constructo 12: Conciencia cognitiva

<p style="text-align: justify">Capacidad de procesamiento consciente de la información, atención sostenida, reflexión metacognitiva y regulación cognitiva.

#### Constructo 13: Conciencia espiritual

<p style="text-align: justify">Capacidad de conexión interna, trascendencia simbólica y percepción de unidad con dimensiones no materiales de la existencia.

#### Constructo 14: Conciencia de propósito

<p style="text-align: justify">Capacidad de orientar la vida hacia metas significativas, proyecto vital y sentido vocacional coherente.

#### Constructo 15: Conciencia integrativa

<p style="text-align: justify">Capacidad de integrar de forma coherente los sistemas cognitivo, emocional, social, corporal, ético y existencial en una estructura unificada del ser.

   Evaluación sistémica del ser

#### 6.2 Escala base del sistema (Likert estandarizada)

Escala principal de medición:

1 = Totalmente en desacuerdo  
2 = En desacuerdo  
3 = Neutral / Indiferente  
4 = De acuerdo  
5 = Totalmente de acuerdo

#### 6.3 Escala de frecuencia

Nunca = 1  
Rara vez = 2  
A veces = 3  
Frecuentemente = 4  
Siempre = 5

#### 6.4 Escala de intensidad

Muy baja = 1  
Baja = 2  
Media = 3  
Alta = 4  
Muy alta = 5

#### 6.5 Sistema de puntuación

Cada conciencia (Ci) se calcula como:

Ci = Σ ítems / n

Donde:
- Ci = puntuación de la conciencia
- Σ ítems = suma de puntuaciones
- n = número de ítems

#### 6.6 Normalización de puntajes

Transformación a escala estándar:

Puntaje normalizado = (X − Min) / (Max − Min)

Resultado en rango:

0.0 – 1.0

### 3. Cálculo de V de Aiken

Para cada ítem se calculará el índice de validez de contenido mediante la fórmula:

V = (Σ s) / [n (c − 1)]

Donde:

- s = puntuación asignada por el juez menos el valor mínimo de la escala  
- n = número de jueces  
- c = número de categorías de respuesta  


**Interpretación sugerida**

0.70 – 0.79 → Ítem revisable  
< 0.70 → Ítem eliminado o reformulado

# Metodología

<p style="text-align: justify">El presente estudio se enmarca en un enfoque cuantitativo, dado que busca medir, analizar e interpretar variables relacionadas con la conciencia humana a través de un instrumento estructurado.</p>

<p style="text-align: justify">El diseño de la investigación es de tipo no experimental, ya que no se manipulan variables independientes, sino que se observan los fenómenos en su contexto natural mediante la recolección de datos a través de cuestionarios.</p>

<p style="text-align: justify">Asimismo, el alcance del estudio es descriptivo-correlacional, debido a que:</p>

- Describe los niveles de desarrollo en las diferentes dimensiones de la conciencia humana.

- Permite identificar relaciones entre dichas dimensiones y los niveles estructurales del modelo (cognitivo, emocional y conductual).

<p style="text-align: justify">La técnica de recolección de información utilizada es la encuesta, implementada mediante un cuestionario estructurado basado en escalas tipo Likert.</p>

<p style="text-align: justify">El instrumento de medición está compuesto por un banco de ítems diseñados para evaluar múltiples dimensiones de la conciencia, los cuales son:</p>

- Normalizados
- Ponderados
- Agrupados por dimensiones
- Integrados en índices compuestos

<p style="text-align: justify">El procesamiento de los datos se realiza mediante:</p>

- Normalización de respuestas
- Cálculo de puntajes ponderados
- Agregación por dimensiones
- Construcción de índices globales
-Reducción estructural a tres niveles (cognición, emoción y conducta)
